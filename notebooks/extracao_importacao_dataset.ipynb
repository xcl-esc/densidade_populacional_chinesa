Importando e fazendo download dos datasets originais.

```python
import os
import requests
import pandas as pd
```

## Célula para extração dos datasets(Maddson GPC e Onu)

```python
DATASETS = {
    "onu_populacao_idade_sexo.xlsx": "https://population.un.org/wpp/assets/Excel%20Files/1_Indicator%20(Standard)/EXCEL_FILES/2_Population/WPP2024_POP_F01_1_POPULATION_SINGLE_AGE_BOTH_SEXES.xlsx",
    "maddison_pib_pc.csv": "https://ourworldindata.org/grapher/gdp-per-capita-maddison-project-database.csv?v=1&csvType=full&useColumnShortNames=false"
}
PASTA = "../data/raw"

os.makedirs(PASTA, exist_ok=True)

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

for nome_arquivo, url in DATASETS.items():
    caminho_completo = os.path.join(PASTA, nome_arquivo)

    try:
        response =  requests.get(url, headers=headers, stream=True, timeout=30)

        if response.status_code == 200:
            with open(caminho_completo, "wb") as f:
                for chunk in response.iter_content(chunk_size=8192):
                    f.write(chunk)

            print(f"Download concluído na pasta: {PASTA}")

        else:
            print(f"Falha no download. Código de status: {response.status_code}")

    except requests.exceptions.Timeout:
        print(f"Tempo limite excedido para o o download do arquivo: {nome_arquivo}.")
    except Exception as e:
        print(f"Erro ao baixar {nome_arquivo}: {e}\n")
```
